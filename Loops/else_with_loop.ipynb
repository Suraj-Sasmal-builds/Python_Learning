#### Else with for loop 

```python
for i in range(1, 6):
    print(i)
else:
    print("Loop Completed!")
```

```python
## for-else with break 

for i in range(1, 6):
    if i == 3:
        break
    print(i)
else:
    print("Loop Completed!")
```

```python
## for-else with continue 

for i in range(1, 6):
    if i  == 3:
        continue
    print(i)
else:
    print("Loop Completed!")
```

```python
## Check whether a number is prime 

num = 7

for i in range(2, num):
    if num % i == 0:
        print("Not a prime number")
        break
else:
    print("Prime number")
```

#### else with while loop 

```python
## Simple while-else 

i = 1

while i <= 5:
    print(i)
    i += 1
else:
    print("Loop completed!")
```

```python
## while-else with break 

i = 1

while i <= 5:
    if i == 3:
        break 
    print(i)
    i += 1
else:
    print("Loop Completed!")
```

```python
## while-else with continue 

i = 0 

while i < 5:
    i += 1

    if i == 3:
        continue 

    print(i)
else:
    print("Loop completed!")
```
