#### While loop 

```python
# Print number from 1 to 5
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
# print number from 1 to 10
i = 1

while i <= 10:
    print(i)
    i += 1
```

```python
## print even numbers from 1 to 20
i = 1

while i <= 20:
    if i % 2 == 0:
        print(i)
    i += 1
```

```python
## Print odd numbers from 1 to 20 

i = 1

while i <= 20:
    if i % 2 != 0:
        print(i)
    i += 1
```

```python
## print multiplication table of 5

i = 1

while i <= 10:
    print("5 x", i, "-", 5 * i)
    i += 1
```

```python
## Sum of first 10 natural numbers 

i = 1

total = 0

while i <= 10:
    total += i
    i += 1
print("Sum =", total)
```

```python
## Countdown timer  

i = 10 

while i >= 1:
    print(i)
    i -= 1
print("Happy New Year!")
```

```python
## Reverese a number 

num = int(input("Enter a number: "))
reverse = 0

while num != 0:
    remainder = num % 10 
    reverse = reverse * 10 + remainder
    num = num // 10 
print(reverse)
```
