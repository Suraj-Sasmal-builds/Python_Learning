#### For Loop 

```python
## Print number from 1 to 10 

for i in range(1,11):
    print(i)
```

```python
## print number from 1 to  100 

for i in range(1, 101):
    print(i)
```

```python
## Print even number from 1 to 20 

for i in range(1,21):
    if i % 2 == 0:
        print(i)
```

```python
## Print odd number from 1 to 20 

for i in range(1,21):
    if i % 2 != 0:
        print(i)
```

```python
## Multiplication Table 

num = int(input("Enter a number: "))

for i in range(1,11):
    print(f"{num} x {i} - {num * i}")
```

```python
## Sum of first 10 numbers 

total = 0

for i in range(1,11):
    total += i
print(total)
```

```python
## Countdown using loop 

for i in range(10,0,-1):
    print(i)
print("Happy New Year")
```

```python
## Print each characterr of a string 

name = "Python"

for i in range(len(name)):
    print(name[i])
```

```python
## Find Factorail of a number 

num = int(input("Enter a number: "))

fact = 1

for i in range(1,num+1):
    fact = fact * i
print(fact)
```

```python
## Nested Loop 

for i in range(1,4):
    for j in range(1,4):
        print(i, j)
```

```python
## Fibonacani series with a loop 

num = int(input("Enter a nuumber: "))

a = 0
b = 1

for i in range(num):
    print(a,end=" ")
    a, b = b, a + b
```
