#### Continue statement with for loop 

```python
# print number from 1 to 10, but skip the 5.

for i in range(1,11):
    if i == 5:
        continue
    print(i)
```

#### Continue with while loop 

```python
i = 0

while i < 10:
    i += 1

    if  i == 5:
        continue
    print(i)
```

```python
### Skip negative numbers using continue 

numbers = [10, -5, 20, -3, 30]

for num in  numbers:
    if num < 0:
        continue 

    print(num)
```
