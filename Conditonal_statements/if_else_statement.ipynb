#### if_else statement

```python
age = 15

if age >= 18:
    print("You are eligble for vote.")
else:
    print("You are not eligble for vote.")
```

```python
age = 45

if age >= 18:
    print("You are eligible for vote.")
else:
    print("You are not eligble for vote.")
```

```python
num = 10

if num % 2 == 0:
    print("Even number.")
else:
    print("Odd number.")
```

```python
num = int(input("Enter a numbe: "))

if num > 0:
    print("Positive number.")
else:
    print("Negative number.")
```

```python
age = int(input("Enter the age of the person: "))

if age >= 18:
    print("You are eligible for driving license.")
else:
    print(f"You are not eliglbe foe driving license.\nYou need to wait for {18 - age} years.")
```

#### Short Hand if-else Statement 

```python
age = 25

result = print("You are eligble for vote.") if age >= 18 else print("You are not eligible for vote.")
```

```python
num = int(input("Enter a number: "))

result = "Even" if num % 2 == 0 else "Odd"

print(result)
```

```python
num = int(input("Enter a number: "))

result = "Positive" if num > 0 else "Negative"

print(result)
```

```python
age = 20 

result = "Eligible" if age >= 18 else "Not eligible"

print(result)
```
