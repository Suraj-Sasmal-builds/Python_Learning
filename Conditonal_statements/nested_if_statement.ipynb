#### elif-statement

```python
marks = int(input("Enter the marks of the student: "))

if marks >= 90:
    print("A")
elif marks >= 80:
    print("B")
elif marks >= 70:
    print("C")
elif marks >= 60:
    print("D")
elif marks >= 50:
    print("E")
else:
    print("F")
```

```python
age = int(input("Enter the age of the person: "))

has_license_text = True 

if age >= 18:

    if has_license_text:
        print("You can drive.")
    else:
        print("You need to pass the driving license.")
else:
    print("You are under age.")
```

```python
num = int(input("Enter the number: "))

if num > 0:
    print("Positive number.")

    if num % 2 == 0:
        print("Even number.")
    else:
        print("Odd number.")
else:
    print("Negative number.")
```
