## Logical Operators 

```python
# AND  Operator 

a = True 
b = False 

print(a and a)
print(a and b)
print(b and a)
print(b and b)
```

```python
# OR Operator 

a = True 
b = False 

print(a or a)
print(a or b)
print(b or a)
print(b or b)
```

```python
# NOT Operator 

print(not a)
print(not b)
```
