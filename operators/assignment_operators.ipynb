#### Assignment Operator 

```python
## = Assign 

a = 5
print(a)
```

```python
# Add and Assign 

a += 10 
print(a)
```

```python
# Subtract  and Assign 

a -= 5
print(a)
```

```python
# Multiply and  Assign 

a *= 5
print(a)
```

```python
# LeftShift and assign 

a <<= 5
print(a)
```

```python
# Rightshift and  assign

a >>= 5
print(a)
```

```python
# Divide and  assign 

a /= 5
print(a)
```

```python
#  FloorDivide and assign 

a //= 2
print(a)
```

```python
# Exponentation and assign 

a **= 5
print(a)
```

```python
# AND and Assign 

b = 5
b &= 3

print(b)
```

```python
# OR and assign

c = 6
c |= 2

print(c)
```

```python
# XOR and assign  

d = 7
d ^= 3

print(d)
```

```python
# NOT and assign 

e = 4

e != 3
print(e)
```
