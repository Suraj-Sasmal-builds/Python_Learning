#### Membership Operators

```python
## in operator 

fruits = ["Apple", "Banana", "Mango", "Orange"]
print("Mango" in fruits)
```

```python
print("Grapes" in fruits)
```

```python
## not in 

fruits = ["Apple", "Banana", "Mango", "Orange"]

print("Grapes" not in fruits)
print("Apple" not in fruits)
```

```python
## membership operators in strings 

name = "Suraj"

print("S" in name)
print("Z" in name)

print("z" not in name)
print("S" not in name)
```

```python
## membership operators in list 

numbers = [10, 20, 30, 40]
print(20 in numbers)
```

```python
## membership operators in Tuple 

values = (100, 200, 300)
print(500 not in values)
```

```python
## membership operator in dictionary

student = {
    "name": "Suraj",
    "age": 20,
    "course": "Python"
}

print("name" in student)
print("Suraj" in student)
```
