#### Arithmetic Operators|

```python
## Addition 

a = 100 
b = 200

print(a + b)
```

```python
## Subtraction 

a = 300 
b = 134

print(a - b)
```

```python
## Multiplication 

a = 34
b = 23

print(a * b)
```

```python
## Division 
a = 23
b = 12

print(a / b)
```

```python
## Floor Division 

a = 243
b = 23

print(a // b)
```

```python
## Modulus 

a = 567
b = 34

print(a % b)
```

```python
## Exponentation 

a = 4
b = 3

print(a ** b)
```

```python
#### Calculator 

a = int(input("Enter a: "))

b = int(input("Enter b: "))

print(f"Addition : {a + b}")
print(f"Subtraction : {a - b}")
print(f"Multiplication : {a * b}")
print(f"Division : {a / b}")
print(f"Floor Division : {a // b}")
print(f"Modulus : {a % b}")
print(f"Exponentation : {a ** b}")
```
