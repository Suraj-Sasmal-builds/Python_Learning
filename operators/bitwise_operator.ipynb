## Bitwise Operator 

```python
## Bitwise AND Operator 

a = 5
b = 3

print(a & b)
```

```python
## Bitwise OR Operator 

a = 5 
b = 3 


print(a | b)
```

```python
## Bitwise XOR Operator 

a = 7
b = 4

print(a ^ b)
```

```python
## Bitwise NOT 

a = 5 
print(~a)

b = 45
print(~b)
```

```python
## Bitwise Left Shift 

a = 4
b = 6

print(a << b)
```

```python
## Bitwise Right Shift 

a = 6
b = 4

print(a >> b)
```
