#### Identity Operators

```python
# is Operator 

a = [10, 20, 30, 40, 50, 60]

b = a 

print(a is b)
```

```python
print(a is not b)
```

```python
# is not 

a = [10, 20, 30]
b = [10, 20, 30]

print(a is not b)
```

```python
print(a is b)
```

```python
a = [10, 20]
b = [10, 20]

print(a == b)
print(a is b)
```
